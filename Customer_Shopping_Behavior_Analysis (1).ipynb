```python
import pandas as pd
```

```python
df = pd.read_csv('customer_shopping_behavior.csv')
df.columns= df.columns.str.strip()
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.strip()
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))

'''df.groupby('Category')-Groups the DataFrame by Category
Example categories: Electronics, Clothing, Books

['Review Rating']-Selects the Review Rating column inside each category

transform(...)-Applies a function to each group

Returns a result with the same length as the original DataFrame

This is why we can assign it back to df['Review Rating']

lambda x: x.fillna(x.median())

For each category:

x = review ratings of that category

x.median() → median rating of that category

fillna(...) → replaces missing values with that median'''
```

```python
df.columns = df.columns.str.lower()
```

```python
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
# create a new column age_group
labels = ['Young Adult','Adult', 'Middle-Aged', 'Senior']
df['age_group']= pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# create column purchase_frequency_days
frequency_mapping = { 
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90}

df['purchase_frequency_days']= df['frequency_of_purchases'].map(frequency_mapping)
```
